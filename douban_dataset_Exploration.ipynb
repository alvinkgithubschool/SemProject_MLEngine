# Douban Dataset Exploration

This notebook explores the movies_cleaned.txt dataset from the douban_dataset(text information) folder.

The Douban dataset contains information about movies including their IDs, names, directors, summaries, and ratings.

```python
# First, lets upload the movies_cleaned.txt file and load it into a dataframe
from google.colab import files
import pandas as pd
import io

# Upload the movies_cleaned.txt file from your local machine
uploaded = files.upload()

# Load the dataset using tab separator and utf-8 encoding
file_name = 'movies_cleaned.txt'
df = pd.read_csv(io.BytesIO(uploaded[file_name]), sep='\t', encoding='utf-8')

# Display the first few rows to verify loading
print('Dataset loaded successfully!')
print('Preview:')
print(df.head())
```

## a. How many rows and columns are contained in the dataset?

```python
# Get the shape of the dataset (rows, columns)
num_rows, num_cols = df.shape

print(f'Answer: The dataset contains {num_rows} rows and {num_cols} columns.')
```

## b. What datatypes are contained in the dataset?

```python
# Get the datatypes of each column
datatypes = df.dtypes

print('Answer: Datatypes in the dataset:')
print(datatypes)
```

## c. Is the dataset complete? i.e. no missing values?

```python
# Check for missing values in the dataset
missing_values = df.isnull().sum()
total_missing = missing_values.sum()

print('Missing values per column:')
print(missing_values)

if total_missing == 0:
    print('Answer: Yes, the dataset is complete. There are no missing values.')
else:
    print(f'Answer: No, the dataset is not complete. There are {total_missing} missing values in total.')
```

## d. Slice out the first 15 rows and last 20 rows and merge them into a new dataframe titled df_sample. Display this new dataframe.

```python
# Slice the first 15 rows
df_first_15 = df.head(15)

# Slice the last 20 rows
df_last_20 = df.tail(20)

# Merge them into a new dataframe titled df_sample
df_sample = pd.concat([df_first_15, df_last_20])

# Display the new dataframe
print('Answer: df_sample:')
df_sample
```
